```python
mov=df[['genres','movie_id','overview','release_date','title_x','cast','crew','keywords']]
mov.head(3)
```

```python
mov['release_date']
```

```python
mov['release_date']=mov['release_date'].fillna('0').apply(lambda x : int(str(x).split('-')[0]))
```

```python
mov['release_date']=mov['release_date'].astype(int)
print(mov[mov['release_date']>0]['release_date'].min(),mov['release_date'].max())
```

```python
def rel_gen(df,col) :
    bins = [1916, 1926, 1936, 1946, 1956, 1966, 1976,1986, 1996, 2006, 2016, 2018]
    label= [f"gen{i}" for i in range(1,12)]
    df[col]=pd.cut(
        df[col], bins=bins, labels=label, right=False
    ).astype('object').fillna('Other')
    return df[col]
```

```python
mov['release_date']=rel_gen(mov,'release_date')
```

```python
mov['release_date'].value_counts()
```

```python
mov[mov['release_date']=='Other']
```

```python
mov['release_date']=mov['release_date'].replace({
    'Other': 'gen8',
    'gen1': 'gen3',
    'gen2': 'gen3',
    'gen3': 'gen4'
})
```

```python
mov.isnull().sum()
```

```python
mov.dropna(inplace=True)
```

```python
mov.duplicated().sum()
```

```python
mov.iloc[0].genres
```

```python
import ast
```

```python
def cov_genres(x) :
    arr=[]
    for i in ast.literal_eval(x) :
        arr.append(i['name'])
    return arr
```

```python
x=cov_genres(mov.iloc[0].genres)
x
```

```python
mov['genres']=mov['genres'].apply(cov_genres)
```

```python
mov['genres']
```

```python
mov.iloc[0].cast
```

```python
def cov_cast(x) :
    obj=ast.literal_eval(x)
    L=[]
    for i in obj[:3] :
        L.append(i['character'])
    return L

```

```python
mov['cast']=mov['cast'].apply(cov_cast)
```

```python
mov.head(3)
```

```python
mov.iloc[2047].crew
```

```python
def cov_crew(x) :
    obj=ast.literal_eval(x)
    L=[]
    for i in obj[:10] :
        if((i['job']=="Director") or i['job']=="Writer") :
            L.append(i['name'])
    return L
```

```python
mov['crew']=mov['crew'].apply(cov_crew)
```

```python
mov.head(3)
```

```python
mov.iloc[0].keywords
```

```python
def cov_keywords(x) :
    L=[]
    obj=ast.literal_eval(x)
    for i in obj[:12] :
        L.append(i['name'])
    return L
```

```python
mov['keywords']=mov['keywords'].apply(cov_keywords)
```

```python
mov.head(3)
```

```python
mov['overview']=mov['overview'].apply(lambda x : x.replace(',','').replace('.','').replace('!','').split())
```

```python
mov.head(3)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
from nltk.corpus import stopwords
stopword=set(stopwords.words('english'))
stemer=PorterStemmer()
```

```python
def normalizer(arr) : 
    output=[]
    for word in arr : 
        word=word.lower()
        if word not in stopword :
            output.append(stemer.stem(word))
    return output
        
```

```python
mov['overview']=mov['overview'].apply(normalizer)
```

```python
mov.head(3)
```

```python
mov['release_date'] = mov['release_date'].apply(lambda x: [x])
mov['tags']=mov['overview']+mov['release_date']+mov['cast']+mov['crew']+mov['keywords']+mov['genres']
```

```python
mov.head(3)
```

```python
dff=mov[['movie_id','title_x','tags']]
dff.head(3)
```

```python
dff['tags']=mov['tags'].replace(' ','')
dff['tags']=mov['tags'].apply(normalizer)
dff.head(3)
```

```python
dff['tags']=dff['tags'].apply(lambda x : ' '.join(x))
dff.head(3)
```

```python
from sklearn.feature_extraction.text import CountVectorizer 
cv=CountVectorizer(max_features=7500)
```

```python
vector=cv.fit_transform(dff['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vector)
similarity.shape
```

```python
def recomend(name):
    matches = dff[dff['title_x'].str.contains(name, case=False, na=False)]
    if matches.empty:
        print("Movie not found!")
        return
    ind = matches.index[0]
    enu = list(enumerate(similarity[ind]))
    res = sorted(enu, reverse=True, key=lambda x: x[1])[1:6]
    for i in res:
        print(dff.iloc[i[0]]['title_x'])
```

```python
recomend('avengers')
```
